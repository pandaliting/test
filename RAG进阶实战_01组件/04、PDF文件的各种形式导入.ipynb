```python
# 使用各种工具解析PDF
# 1、使用PyPDFLoader加载文件
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("../asset/load/04-sample.pdf")
pages = loader.load()
print(len(pages))
for page in pages:
    print(page.page_content)
```

```python
# 2、使用MyMuPDF工具
import fitz  # PyMuPDF

doc = fitz.open("../asset/load/04-sample.pdf")

for page in doc:
    text = page.get_text()
    print(text)

doc.close()
```
