```python
# 整体原则
# 优先使用整体工具，理解文件的内部结构，将复杂的文件简单化
# 导入text文件，加载进内存
from langchain_community.document_loaders import TextLoader

loader = TextLoader(
    file_path="../asset/load/01-langchain-utf-8.txt",
    encoding="utf-8"
)
doc = loader.load()
print(doc)

# metadata 中可以添加其他信息
```

```python
# 读取目录中所有的txt文档
from langchain_community.document_loaders import DirectoryLoader

loader_dir = DirectoryLoader(
    path="../asset/load",
    glob="*.txt",
    loader_cls=TextLoader,
    recursive=True,
    show_progress=True,
)

docs_dir = loader_dir.load()
print(len(docs_dir))
print(docs_dir)
```

```python
# 解析markdown文件

loader_md = DirectoryLoader(
    path="../asset/load",
    glob="*.md",
    loader_cls=TextLoader,
    recursive=True,
    show_progress=True,
)
docs_md = loader_md.load()
print(docs_md)
```
