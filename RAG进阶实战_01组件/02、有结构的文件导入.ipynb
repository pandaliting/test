```python
# 有结构的文件导入
# 加载JSON文件
from langchain_community.document_loaders import JSONLoader

load_json = JSONLoader(
    file_path='../asset/load/03-load.json',
    jq_schema='.messages[]',
    text_content=False,
)
doc_json = load_json.load()
print(doc_json)
```

```python
# 网页文档的解析
from langchain_community.document_loaders import WebBaseLoader

# 单个 URL
loader = WebBaseLoader(
    web_paths=["https://zh.wikipedia.org/wiki/黑神话：悟空"],
)
docs = loader.load()
print(f"共加载 {len(docs)} 个页面")
# print(docs)
print(docs[0].metadata)
```

```python
# 使用unstructuredloader 加载网页
from langchain_community.document_loaders import UnstructuredURLLoader

urls = [
    "https://zh.wikipedia.org/wiki/黑神话：悟空",
]

loader = UnstructuredURLLoader(
    urls=urls,
    mode="elements",  # 按元素切分（标题、段落、列表等）
    show_progress_bar=True,
)
docs = loader.load()
print(f"共加载 {len(docs)} 个元素")
for doc in docs[:5]:
    print(f"[{doc.metadata.get('category', '')}] {doc.page_content[:80]}")
```

```python
# 导入解析markdown文件
from langchain_community.document_loaders import DirectoryLoader, TextLoader
from langchain_text_splitters import MarkdownHeaderTextSplitter

load_md = TextLoader("../asset/load/06-load.md", encoding="utf-8")
doc_md = load_md.load()
# print(doc_md)
# print(doc_md[0])
# 切分
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#", "H1"),
        ("##", "H2"),
        ("###", "H3"),
    ]
)
chunks = splitter.split_text(doc_md[0].page_content)
for chunk in chunks[:3]:
    print(f"标题: {chunk.metadata}")
    print(f"内容: {chunk.page_content[:100]}")
    print("---")
```
