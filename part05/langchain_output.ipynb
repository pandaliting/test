```python
# 结构化输出
from pydantic import BaseModel, Field
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv(override=True)

qwen_chat = ChatOpenAI(
    model="qwen3.7-max",
    api_key=os.getenv("DASHSCOPE_API_KEY"),
    base_url=os.getenv("DASHSCOPE_BASE_URL"),
    extra_body={"enable_thinking": False},
)


# 定义Pydantic 模型
class Person(BaseModel):
    name: str = Field(description="姓名")
    age: int = Field(description="年龄")
    occupation: str = Field(description="职业")


# 内部把schema包装成tool传递给大模型
structured_llm = qwen_chat.with_structured_output(Person)
# 大模型返回+解析校验
result = structured_llm.invoke("张三是一个30岁的工程师")
print(result.name)
# 一句话总结：with_structured_output 帮你做了 JSON Schema 转换 → Tool 调用 → 解析校验 全部自动化，你只管定义结构、调用、取字段。
```
